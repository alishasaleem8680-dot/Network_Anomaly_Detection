```python
# Remove duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

print("Duplicates removed successfully!")
print("New dataset shape:", df.shape)
print("Remaining duplicates:", df.duplicated().sum())
```

```python
# Separate features and ground-truth labels

X = df.drop(columns=["Attack Type"])
y = df["Attack Type"]

print("Features shape:", X.shape)
print("Label shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())

print("\nLabel distribution:")
print(y.value_counts())
```

```python
# Check data types of all features

print(X.dtypes.value_counts())

print("\nNon-numeric columns:")
print(X.select_dtypes(exclude=np.number).columns.tolist())
```

```python
# Check columns with only one unique value

constant_columns = X.columns[X.nunique() <= 1].tolist()

print("Constant columns:")
print(constant_columns)

print("\nNumber of constant columns:", len(constant_columns))
```

```python
# Convert features to float32 to reduce memory usage

X = X.astype(np.float32)

print("Data type conversion completed!")
print(X.dtypes.value_counts())
print("X shape:", X.shape)
```

```python
print("Memory usage of X:")
print(f"{X.memory_usage(deep=True).sum() / (1024**2):.2f} MB")
```

```python
# Create a manageable training sample for Isolation Forest

X_train = X.sample(n=200_000, random_state=42)

print("Training sample created successfully!")
print("Training data shape:", X_train.shape)
```

```python
import sklearn

print("scikit-learn installed successfully!")
print("Version:", sklearn.__version__)
```

```python
from sklearn.ensemble import IsolationForest

# Create Isolation Forest model
model = IsolationForest(
    n_estimators=100,
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

# Train the model
model.fit(X_train)

print("Isolation Forest trained successfully!")
```

```python
# Create evaluation sample

eval_df = df.sample(n=100_000, random_state=123)

X_test = eval_df.drop(columns=["Attack Type"]).astype(np.float32)
y_test = eval_df["Attack Type"]

print("Test data shape:", X_test.shape)
print("Actual labels shape:", y_test.shape)
```

```python
# Predict anomalies

predictions = model.predict(X_test)

print("Prediction completed!")
print("Unique predictions:", np.unique(predictions))
```

```python
# Convert Isolation Forest predictions into readable labels

y_pred = np.where(predictions == -1, "Anomaly", "Normal")

print("Prediction labels created successfully!")
print(pd.Series(y_pred).value_counts())
```

```python
# Convert actual labels into Normal vs Attack

y_true = np.where(
    y_test == "Normal Traffic",
    "Normal",
    "Attack"
)

print("Actual labels converted successfully!")
print(pd.Series(y_true).value_counts())
```

```python
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# Convert labels to binary values
# Normal = 0
# Attack/Anomaly = 1

y_true_binary = np.where(y_true == "Attack", 1, 0)
y_pred_binary = np.where(y_pred == "Anomaly", 1, 0)

# Confusion Matrix
cm = confusion_matrix(y_true_binary, y_pred_binary)

print("Confusion Matrix:")
print(cm)

# Metrics
accuracy = accuracy_score(y_true_binary, y_pred_binary)
precision = precision_score(y_true_binary, y_pred_binary, zero_division=0)
recall = recall_score(y_true_binary, y_pred_binary, zero_division=0)
f1 = f1_score(y_true_binary, y_pred_binary, zero_division=0)

print("\nModel Performance:")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
```

```python
# Calculate anomaly scores

anomaly_scores = model.decision_function(X_test)

print("Anomaly scores calculated successfully!")
print("Minimum score:", anomaly_scores.min())
print("Maximum score:", anomaly_scores.max())
```
